```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
from openai.types.chat import ChatCompletionUserMessageParam
from openai.types.chat import ChatCompletionMessageParam
```

```python
load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
    OpenAI.api_key = api_key
```

```python
system_prompt = """
You are a customer support ticket classification assistant.

Analyze the ticket and determine:
- Category
- Priority
- Customer sentiment
- Main issue
- Recommended next action

Priority must be one of:
- Low
- Medium
- High
- Critical

Do not invent information.
Respond in markdown.
"""
```

```python
user_prompt = """
Analyze this support ticket:

Customer: ABC Trading

Our users cannot log into the CRM since this morning.
The login page loads correctly, but clicking the login button
returns a 500 error.

This is affecting all 25 sales representatives.
"""
```

```python
messages: list[ChatCompletionMessageParam] = [
    {
        "role": "system",
        "content": system_prompt,
    },
    {
        "role": "user",
        "content": user_prompt,
    }
]
```

```python
client = OpenAI()
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)
content = response.choices[0].message.content
display(Markdown(content))
```

```python
system_CRM_email_prompt = """
You are an AI sales email assistant for a CRM system.

Your task is to help sales executives write professional,
customer-friendly emails based on CRM context.

Write emails that are:
- Professional
- Polite
- Clear
- Concise
- Natural
- Relevant to the customer's requirement
- Appropriate for a business sales conversation

Use the provided subject as the primary communication intent.

Use CRM information such as the lead title, customer name,
company, sales stage, and customer requirement to add relevant
context when appropriate.

Do not invent:
- Product specifications
- Prices
- Discounts
- Delivery dates
- Commitments
- Customer requirements
- Previous conversations
- Business facts

If important information is missing, write the email without
inventing it.

Do not repeat the subject as the first sentence unnecessarily.

The email should normally contain:
- A professional greeting
- A short opening relevant to the customer's context
- The main purpose of the email
- A clear next step or call to action when appropriate
- A professional closing

Keep the email concise and suitable for a sales executive.

Return only the email body in markdown.
"""
```

```python
user_CRM_email_prompt = """
Write a professional sales email using the following CRM context.

Recipient:
Lathusa

Company:
ABC Trading

Lead title:
need 10 laptops

Subject:
Follow-up regarding your laptop requirement

Sales stage:
New

Expected revenue:
50000

The email should acknowledge the customer's requirement,
show willingness to help, and ask for the required laptop
specifications so that the sales team can recommend suitable
options and prepare the next step.
"""
```

```python
crm_email_messages: list[ChatCompletionMessageParam] = [
    {
        "role": "system",
        "content": system_CRM_email_prompt,
    },
    {
        "role": "user",
        "content": user_CRM_email_prompt,
    }
]
```

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=crm_email_messages
)
content = response.choices[0].message.content
display(Markdown(content))
```
